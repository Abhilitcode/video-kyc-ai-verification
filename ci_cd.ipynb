## 1. What is CI/CD?

**CI/CD = Continuous Integration / Continuous Delivery or Continuous Deployment.**

Simple idea:

> **Automatically check, build, and potentially deploy your application whenever code changes.**

Instead of:

```text
Code → manually test → manually deploy
```

we want:

```text
Code
 ↓
Git push
 ↓
CI/CD pipeline
 ↓
Test
 ↓
Build
 ↓
Deploy
```

---

# 2. CI — Continuous Integration

### Easy definition

> **CI means automatically testing and validating code whenever developers integrate/push changes into a shared repository.**

Example:

```text
Developer
   ↓
git push
   ↓
GitHub
   ↓
Run tests
   ↓
PASS ✅ / FAIL ❌
```

### Why?

Imagine 5 developers are working on the same project.

Without CI:

```text
Everyone writes code
       ↓
Everything merged
       ↓
💥 Something breaks
```

With CI:

```text
Developer pushes code
       ↓
Automated checks
       ↓
Problem detected early
```

### Interview definition

> **"Continuous Integration is the practice of automatically building and testing code changes whenever they are integrated into a shared repository."**

---

# 3. CD — Continuous Delivery vs Continuous Deployment

This is where people sometimes get confused.

## Continuous Delivery

Code is automatically:

```text
tested
 ↓
built
 ↓
prepared for deployment
```

but a **human may approve the final deployment**.

```text
Code
 ↓
CI
 ↓
Build
 ↓
Ready to deploy
 ↓
👤 Approval
 ↓
Production
```

---

## Continuous Deployment

Everything is automated:

```text
Code
 ↓
Test
 ↓
Build
 ↓
Deploy automatically
 ↓
Production
```

No manual approval for every deployment.

### Interview distinction

> **Continuous Delivery → always ready to deploy, often with a manual release decision.**

> **Continuous Deployment → successful changes are automatically deployed.**

---

# 4. What is a Pipeline?

A **pipeline** is the sequence of automated stages that code goes through.

For your project:

```text
Git Push
   ↓
Install dependencies
   ↓
Run tests
   ↓
Build/check
   ↓
Deploy
```

That's your **CI/CD pipeline**.

Think of it like an assembly line.

---

# 5. GitHub Actions

We're going to use **GitHub Actions**.

### What is it?

> **GitHub Actions is GitHub's automation/CI/CD platform that lets you automatically run workflows when events happen in your repository.**

For example:

```text
git push
   ↓
GitHub Actions starts
```

It can:

* install Python
* install dependencies
* run tests
* lint code
* build Docker images
* deploy applications

---

# 6. Workflow

A **workflow** is the YAML file that tells GitHub Actions what to do.

Usually:

```text
.github/
└── workflows/
    └── ci-cd.yml
```

The YAML describes:

```text
WHEN should it run?
WHAT should it run?
WHERE should it run?
```

---

# 7. Trigger

A **trigger** determines **when the workflow starts**.

Example:

```yaml
on:
  push:
    branches:
      - main
```

Meaning:

> Run this workflow whenever code is pushed to `main`.

Other common triggers:

```text
push
pull_request
workflow_dispatch
schedule
```

### `workflow_dispatch`

This means:

> Let me manually start the workflow from GitHub.

Very useful while testing.

---

# 8. Runner

A **runner is the machine that executes your GitHub Actions workflow.**

For example:

```yaml
runs-on: ubuntu-latest
```

means:

> Run this workflow on a GitHub-hosted Ubuntu machine.

Think:

```text
GitHub
   ↓
"Hey runner, execute these commands."
   ↓
Temporary machine
   ↓
Run tests/build/etc.
```

---

# 9. Job

A **job is a group of steps that execute together on a runner.**

Example:

```yaml
jobs:
  test:
    runs-on: ubuntu-latest
```

You could have:

```text
Jobs

├── test
├── build
└── deploy
```

Jobs can also depend on each other.

For example:

```text
test
 ↓
build
 ↓
deploy
```

---

# 10. Step

A **step is an individual action/command inside a job.**

Example:

```yaml
steps:
  - uses: actions/checkout@v4

  - name: Install dependencies
    run: pip install -r requirements.txt

  - name: Run tests
    run: pytest
```

So:

```text
Workflow
   ↓
Job
   ↓
Steps
```

Remember this hierarchy:

> **Workflow → Job → Step**

---

# 11. `uses` vs `run`

You'll see these constantly.

### `uses`

Uses an existing GitHub Action.

Example:

```yaml
- uses: actions/checkout@v4
```

This checks your repository's code into the runner.

### `run`

Runs a shell command.

Example:

```yaml
- run: pip install -r requirements.txt
```

So:

> **`uses` → use a predefined action**

> **`run` → execute a command**

---

# 12. Secrets

Never put credentials directly into your workflow.

❌ Bad:

```yaml
OPENAI_API_KEY: "actual-secret"
```

Instead use GitHub Secrets:

```text
GitHub Repository
   ↓
Settings
   ↓
Secrets and variables
   ↓
Actions
   ↓
OPENAI_API_KEY
```

Then the workflow can access it securely.

Example:

```yaml
env:
  OPENAI_API_KEY: ${{ secrets.OPENAI_API_KEY }}
```

The important concept:

> **Secrets allow sensitive credentials to be provided to the pipeline without hardcoding them in the repository.**

---

# 13. Environment

An environment represents a deployment target/configuration.

Common example:

```text
Development
     ↓
Staging
     ↓
Production
```

For a larger application you might have:

```text
dev → test changes

staging → test production-like setup

production → real users
```

You don't necessarily need all three for your KYC project.

---

# 14. Build

A **build** converts your source code into something that can be run/deployed.

For Python:

```text
Source code
 ↓
Install dependencies
 ↓
Run checks/tests
 ↓
Package/application ready
```

For Docker:

```text
Dockerfile
 ↓
docker build
 ↓
Docker image
```

---

# 15. Artifact

An **artifact is a file/output produced by a workflow that you may want to keep or pass to another stage.**

Examples:

```text
test report
build package
Docker image
compiled application
```

For your current Streamlit project, we may not need to make artifacts a major part of the pipeline.

# 16. CI/CD for YOUR KYC project

Our eventual pipeline could be:

```text
                GitHub
                   │
                git push
                   ↓
          GitHub Actions
                   │
             ┌─────┴─────┐
             ↓            ↓
        Install deps    Checkout
             │
             ↓
          Run tests
             │
          ┌──┴──┐
          ↓     ↓
       PASS    FAIL
          │
          ↓
     Build/check
          │
          ↓
       Deploy
```

And because your project is already on **Streamlit Cloud**, we need to decide how much CI/CD we actually need.

We don't need to invent a complicated AWS deployment pipeline just to say we have CI/CD.

# 17. CI vs CD — easiest way to remember

### CI

> **"Does my new code work?"**

```text
Push
 ↓
Test
 ↓
PASS / FAIL
```

### CD

> **"Get my working code deployed."**

```text
Test
 ↓
Build
 ↓
Deploy
```

---

# 18. Important interview concepts

### What happens if tests fail?

Usually:

```text
Test ❌
 ↓
Pipeline stops
 ↓
Deployment doesn't happen
```

This prevents broken code from reaching production.

### Why automate CI/CD?

Because it gives:

* faster feedback
* repeatable deployments
* fewer manual mistakes
* consistent testing
* safer releases

---

# 19. Your interview answer

If asked:

### "What is CI/CD?"

Say:

> **"CI/CD is a software development practice where code changes are automatically tested, validated, built and, depending on the setup, deployed. CI focuses on continuously integrating and testing changes, while CD focuses on reliably delivering or automatically deploying those changes."**

### "What is GitHub Actions?"

> **"GitHub Actions is GitHub's automation platform that allows us to define workflows which run automatically based on events such as pushes or pull requests. We can use it to run tests, build applications and deploy them."**

### "What is a runner?"

> **"A runner is the machine or environment where GitHub Actions executes the jobs and commands defined in a workflow."**

### "Workflow vs job vs step?"

> **"A workflow is the overall automation process, a job is a group of related tasks running on a runner, and a step is an individual command or action within a job."**
