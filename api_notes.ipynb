# 2. What is an API?

API = **Application Programming Interface**.

In very simple words:

> **An API is a way for one program to communicate with another program.**

For example, your frontend can say:

> "Here is a user's photo and video. Please perform KYC."

```json
{
  "decision": "REVIEW",
  "reason": "Liveness evidence was insufficient"
}
```

# 6. HTTP Status Codes

This is another **must-know interview topic**.

The server doesn't only return data. It also tells the client whether the request succeeded.

### `200 OK`

Everything worked.

```text
POST /verify-kyc
→ 200 OK
```

### `201 Created`

Something was successfully created.

For example:

```text
POST /users
→ 201 Created
```

### `400 Bad Request`

The client sent invalid input.

Example:

```text
Photo missing
→ 400 Bad Request
```

### `401 Unauthorized`

Authentication is missing/invalid.

```text
No valid authentication
→ 401
```

### `403 Forbidden`

The client is authenticated but **doesn't have permission**.

```text
User authenticated
BUT
doesn't have access
→ 403
```

### `404 Not Found`

Requested resource doesn't exist.

```text
GET /users/999999
→ 404
```

### `500 Internal Server Error`

Something went wrong on the server.

```text
Unexpected backend failure
→ 500
```

## Request Body + Pydantic

When an API needs to receive **structured information**, we usually put that information in the **request body**.

For example, suppose we have a user registration API.

### Client sends:

```json
{
    "name": "Abhishek",
    "email": "abhishek@example.com",
    "age": 24
}
```

This JSON is the **request body**.

---

# 1. What is Pydantic?

**Pydantic** is commonly used with FastAPI to **validate and structure incoming data**.

Think of it as a **checker**.

You tell it:

```text
name → must be a string
email → must be a string
age → must be an integer
```

Then FastAPI/Pydantic checks the incoming request.

---

## 2. Example

```python
from fastapi import FastAPI
from pydantic import BaseModel

app = FastAPI()


class User(BaseModel):
    name: str
    email: str
    age: int


@app.post("/users")
def create_user(user: User):
    return {
        "message": "User created",
        "name": user.name
    }
```

Let's break this down.

### This:

```python
class User(BaseModel):
```

creates a **Pydantic model**.

We're defining what our input should look like.

Then:

```python
name: str
email: str
age: int
```

means:

```text
name  → string
email → string
age   → integer
```

Then:

```python
def create_user(user: User):
```

means:

> FastAPI expects the request body to follow the `User` structure.

---

# 3. Valid request

Client sends:

```json
{
    "name": "Abhishek",
    "email": "abhishek@example.com",
    "age": 24
}
```

✅ Valid.

FastAPI passes it to:

```python
create_user(user)
```

# 4. Invalid request

Suppose someone sends:

```json
{
    "name": "Abhishek",
    "email": "abhishek@example.com",
    "age": "hello"
}
```

But we said:

Pydantic detects that the data doesn't match the expected structure.

FastAPI returns a validation error instead of letting bad data continue into our backend logic.

# 1. Authentication vs Authorization

These are often confused.

### Authentication = "Who are you?"

It verifies your identity.

Example:

```text
User → username + password
     → authentication
     → "Yes, this is Abhishek."
```

Common mechanisms:

* API keys
* JWT tokens
* OAuth
* username/password

---

### Authorization = "What are you allowed to do?"

After we know who you are, we check your permissions.

Example:

```text
Abhishek
   ↓
Authenticated ✅
   ↓
Can access /verify-kyc? ✅
Can access /admin/users? ❌
```

So remember:

> **Authentication → Who are you?**

> **Authorization → What can you access?**

You already encountered this with AWS IAM:

```text
IAM
 ↓
Identity + permissions
```

# 2. Async in FastAPI

You will see:

```python
async def verify_kyc():
    ...
```

instead of:

```python
def verify_kyc():
    ...
```

### What's the idea?

Imagine your server receives a request and then waits for something external:

```text
Request
  ↓
Call API
  ↓
WAIT................
  ↓
Response
```

With normal synchronous execution, that worker may remain occupied while waiting.

With asynchronous programming, the server can potentially work on other tasks while waiting for an I/O operation.

Simple analogy:

### Synchronous

You're at a restaurant and:

> Order food → stand at the kitchen → wait → receive food.

### Asynchronous

> Order food → sit down → do something else → food arrives.

---

### Important:

**Async does NOT automatically make every program faster.**

It's particularly useful for **I/O-bound work** such as:

* API calls
* database operations
* network requests
* file operations

For CPU-heavy work like certain computer-vision processing, simply adding `async` doesn't magically make it faster.

For your KYC pipeline, we don't need to convert everything to async just for the sake of it.
