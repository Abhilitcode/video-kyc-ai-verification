
# 1. Observability — easiest definition

Think of your KYC application like a car.

### Monitoring asks:

> 🚨 Is something wrong?

For example:

```text
Average KYC latency = 35 seconds
KYC errors = 5
```

You know **something is happening**.

### Observability asks:

> 🔍 Why is it happening?

For example:

```text
KYC took 52 seconds

S3 upload       → 5 sec
OCR             → 2 sec
Face            → 1 sec
Liveness        → 30 sec  ← bottleneck
RAG             → 2 sec
LLM             → 12 sec
```

Now you can actually investigate the problem.

---

# 2. The 3 pillars of Observability

This is the **must-know interview concept**.

## ① Logs 📝

Logs tell you **what happened**.

You already implemented this.

Example:

```python
logger.info("KYC verification started")
```

or:

```text
INFO - KYC final decision recorded as REVIEW
```

Logs are detailed events.

---

## ② Metrics 📊

Metrics tell you **how much / how often** something happened.

You already implemented this with CloudWatch.

For example:

```text
TotalLatency = 35 sec
LivenessLatency = 13 sec
LLMLatency = 11 sec

KYC_PASS = 1
KYC_REVIEW = 1
KYC_Errors = 0
```

Metrics are numbers that are easy to aggregate and visualize.

---

## ③ Traces 🔗

This is the **new part for us**.

A trace tells you the **complete journey of one request through the system**.

For your KYC:

```text
KYC Request
     │
     ├── S3 Upload
     │
     ├── OCR
     │
     ├── Face Verification
     │
     ├── Liveness
     │
     ├── RAG
     │
     └── LLM
```

Think of it like putting a **tracking number on one KYC request** and following it through every stage.

# 3. Monitoring vs Observability

This distinction is important in interviews.

### Monitoring

> **"Tell me when something is wrong."**

Example:

```text
LivenessLatency > 30 sec
```

CloudWatch can alert you.

### Observability

> **"Give me enough information to understand why it is wrong."**

You might discover:

```text
Request #1234

S3       → 4 sec
OCR      → 2 sec
Face     → 1 sec
Liveness → 38 sec  ← problem
RAG      → 1 sec
LLM      → 11 sec
```

So:

**Monitoring detects the problem.**
**Observability helps investigate the problem.**

That's a very good interview answer.

### 1. What is a Trace?

A **trace = the complete journey of one request through your application.**

For your KYC system, imagine one employee submits:

> Photo ID + video → KYC decision

That entire journey is **one trace**.

```text
Trace ID: ABC123

KYC Request
    │
    ├── S3 Upload
    ├── OCR
    ├── Face Verification
    ├── Liveness
    ├── RAG
    └── LLM
```

So remember:

> **Trace = complete request journey**

---

### 2. What is a Span?

A **span = one individual operation inside that trace.**

For example:

```text
Trace: ABC123
│
├── Span: S3 Upload
├── Span: OCR
├── Span: Face Verification
├── Span: Liveness
├── Span: RAG
└── Span: LLM
```

Each stage gets its own span.

A span can contain information such as:

```text
Operation: Liveness
Start: 10:30:01
End:   10:30:14
Duration: 13 sec
Status: SUCCESS
```

---

### 3. Easy analogy 🏥

Imagine going to a hospital.

**Your entire hospital visit = Trace**

```text
Hospital Visit
   │
   ├── Registration
   ├── Doctor consultation
   ├── Blood test
   ├── X-ray
   └── Pharmacy
```

Each individual activity is a **Span**.

So:

> **Trace = whole journey**
> **Span = one step in the journey**

---

## 4. Parent and Child Spans

Spans can have relationships.

```text
Trace
│
└── KYC Verification       ← parent span
      │
      ├── S3 Upload        ← child
      ├── OCR               ← child
      ├── Face              ← child
      ├── Liveness          ← child
      ├── RAG               ← child
      └── LLM               ← child
```

You don't necessarily need to create a giant complicated hierarchy for your project. The important interview concept is simply:

> A **parent span** represents a broader operation, while **child spans** represent operations performed inside it.

# 5. Why do we need this?

Suppose CloudWatch says:

```text
TotalLatency = 50 seconds
```

That's monitoring.

But **why 50 seconds?**

A trace can show:

```text
S3       → 5 sec
OCR      → 2 sec
Face     → 1 sec
Liveness → 29 sec  🔴
RAG      → 2 sec
LLM      → 11 sec
```

Now you immediately know:

> **Liveness is responsible for most of the latency.**

That's the power of tracing.

# 6. Trace ID vs Span ID

Another interview favourite.

### Trace ID

Identifies the **entire request**.

```text
Trace ID = ABC123
```

Every span belonging to that KYC request is associated with `ABC123`.

### Span ID

Identifies **one particular operation**.

```text
Trace ID: ABC123

S3        → Span ID: 001
OCR       → Span ID: 002
Face      → Span ID: 003
Liveness  → Span ID: 004
RAG       → Span ID: 005
LLM       → Span ID: 006
```

So:

> **Trace ID → identifies the whole request**
> **Span ID → identifies one operation**

# 7. The one-line interview answer

If interviewer asks:

**"What is the difference between a trace and a span?"**

Say:

> **"A trace represents the complete journey of a request through a system, while a span represents an individual operation within that trace. For example, in my KYC system, the entire verification is a trace and OCR, face verification, liveness, RAG, and LLM processing can each be represented as spans."**



The difference is not about the data you collect. The difference is what you do with it.Monitoring uses metrics and logs to watch for known problems. You set up a dashboard to watch things you already expect to break.Observability uses metrics, logs, and traces together. It connects them so you can investigate strange, new problems that you have never seen before.


## 1. OpenTelemetry — what is it?

**OpenTelemetry (OTel)** is an open-source standard/toolset used to **collect observability data** from applications.

That data mainly includes:

* **Traces** 🔗
* **Metrics** 📊
* **Logs** 📝

Think of it as a **common language for observability data**.

Instead of an application having completely different ways to generate telemetry for every monitoring platform, OpenTelemetry provides standard instrumentation and data formats.

---

## 2. Where does it fit?

Your current project:

```text
                    KYC APPLICATION
                          │
          ┌───────────────┼───────────────┐
          ↓               ↓               ↓
        Logs           Metrics          Timing
          │               │               │
     Watchtower       CloudWatch      perf_counter
          │               │
          └───────────────┘
                  ↓
              CloudWatch
```

If we **had** used OpenTelemetry:

```text
                    KYC APPLICATION
                          │
                    OpenTelemetry
                          │
             ┌────────────┼────────────┐
             ↓            ↓            ↓
           Logs        Metrics       Traces
             │            │            │
             └────────────┼────────────┘
                          ↓
                  Observability backend
```

OTel would sit between the application and the observability system.

---

# 3. Why do companies use OpenTelemetry?

Imagine a company initially uses:

```text
Application → CloudWatch
```

Later they want:

```text
Application → Datadog
```

or:

```text
Application → Grafana
```

or another observability backend.

OpenTelemetry helps standardize the telemetry collection so the application isn't tightly tied to one vendor.

That's one of its major benefits.

# 4. OpenTelemetry vs CloudWatch

This distinction is important.

**CloudWatch** is an AWS monitoring/observability service.

**OpenTelemetry** is an observability instrumentation/standard framework.

So they're not exactly competitors.

You could have:

```text
Python Application
       ↓
OpenTelemetry
       ↓
AWS / CloudWatch
```

Your project instead uses:

```text
Python Application
       ↓
Boto3 / Watchtower / logging
       ↓
CloudWatch
```

Both approaches are possible.

# 5. What would a Trace look like?

For your KYC application:

```text
Trace ID: ABC123
│
└── KYC Verification
      │
      ├── S3 Upload
      │
      ├── OCR
      │
      ├── Face Verification
      │
      ├── Liveness
      │
      ├── RAG
      │
      └── LLM
```

Each operation could be represented as a **span**.

For example:

```text
Liveness Span
├── Start time
├── End time
├── Duration
├── Status
└── Attributes
```

So if a request takes 50 seconds, tracing lets you see **where that time was spent**.
